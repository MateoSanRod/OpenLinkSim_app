```python
import pandas as pd
```

```python
file_path = r"C:\Users\teoto\Downloads\PMKS+ Desktop\New folder\Kinematic Data from PMKSD  22_35_18_23_11_2024.txt".replace('\\', '/')
df = pd.read_csv(file_path, delimiter='\t', encoding='utf-16')
df
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['font.family'], plt.rcParams['mathtext.rm'], plt.rcParams['mathtext.it'], \
    plt.rcParams['mathtext.bf'], plt.rcParams['mathtext.sf'] = 5 * ['Times New Roman']

# Set up the figure and axis
fig, ax1 = plt.subplots(figsize=(10, 4))
plt.rcParams['hatch.linewidth'] = 3

# Plot for df['column1'] on the left y-axis
ax1.plot(df['TimeSteps'], df['x 3,4']/100,
         color='tab:blue',
         linestyle='--',
         label='Slider displacement', linewidth=1)
ax1.set_ylabel(r'Displacement [$m$]', color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')

# Create a second y-axis for velocity
ax2 = ax1.twinx()
ax2.plot(df['TimeSteps'], df['Vx 3,4']/100,
         color='tab:red',
         linestyle='--',
         label='Slider velocity', linewidth=1)
ax2.set_ylabel('Velocity [$m/s$]', color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')

# Create a third y-axis for acceleration
ax3 = ax1.twinx()
ax3.spines['right'].set_position(('outward', 60))  # Offset the third axis
ax3.plot(df['TimeSteps'], df['Ax 3,4']/100,
         color='tab:green',
         linestyle='--',
         label='Slider acceleration', linewidth=1)
ax3.set_ylabel('Acceleration [$m/s^2$]', color='tab:green')
ax3.tick_params(axis='y', labelcolor='tab:green')

ax1.set_xticks(np.arange(0, 2 + .1, 0.5))
ax1.set_xticks(np.arange(0, 2 + .1, 0.25), minor=True)
ax1.set_xlim(0, 2)

# Additional settings for the x-axis
ax1.set_xlabel('Time (s)')

# Show the legend for all axes
fig.legend(loc='upper center', bbox_to_anchor=(0.5, 1), ncol=7)

# Save or display the plot
plt.savefig('multi_column_plot.png', dpi=300)
# Adjust layout to prevent clipping when saving
fig.tight_layout(rect=[0, 0, 1, 0.95])

# Save or display the plot
plt.savefig('multi_column_plot.png', dpi=300)
```
